# Control Databases

## Bonds

### NaNs check

```python
import sys
import os
import pandas as pd


bonds = pd.read_parquet(r'C:\Users\benel\Coding\Python\Database\bonds.parquet')

dates = bonds.index.get_level_values('Date')
print(f'Date range: {dates.min().date()} to {dates.max().date()}')
nan_mask = bonds.isna().any(axis=1)
print(f'Rows with at least one NaN: ')

nan_by_ticker = (
    bonds[nan_mask]
    .index.to_frame(index=False)
    .groupby('Ticker')['Date']
    .agg(count='count', first='min', last='max')
)
print('\nNaN occurrences by ticker:')
print(nan_by_ticker)

bonds
```

### Unique Bonds

```python
tickers = bonds.index.get_level_values('Ticker')
print(tickers.unique())
print(tickers.value_counts())
```

### Visualize Bonds

```python
import matplotlib.pyplot as plt

ticker = '^TNX'
data  = 'Yield'         # Yield for Indices, Close for ETFs

name = bonds["Name"].xs(ticker, level="Ticker").iloc[0]

plt.figure(figsize=(12,6))

plt.plot(bonds[data].xs(ticker, level="Ticker"))
plt.title(name)
plt.grid(True)

plt.show()
```

## Equities

### NaNs check

```python
import sys
import os
import pandas as pd


equities = pd.read_parquet(r'C:\Users\benel\Coding\Python\Database\equities.parquet')

dates = equities.index.get_level_values('Date')
print(f'Date range: {dates.min().date()} to {dates.max().date()}')
nan_mask = equities.isna().any(axis=1)
print(f'Rows with at least one NaN: ')

nan_by_ticker = (
    equities[nan_mask]
    .index.to_frame(index=False)
    .groupby('Ticker')['Date']
    .agg(count='count', first='min', last='max')
)
print('\nNaN occurrences by ticker:')
print(nan_by_ticker)

equities
```

### Unique Tickers

```python
tickers = equities.index.get_level_values('Ticker')
print(tickers.unique())
print(tickers.value_counts())
```

### Visualize Equities

```python
import matplotlib.pyplot as plt

ticker = 'META'
data  = 'Close'         # Yield for Indices, Close for ETFs

equities_mapped = pd.read_parquet(r'C:\Users\benel\Coding\Python\Database\equities_mapping.parquet')
name = equities_mapped.loc[ticker, 'Name']

fig, (ax_price, ax_vol) = plt.subplots(
    2, 1, figsize=(12, 8), sharex=True, height_ratios=[3, 1]
)

ax_price.plot(equities[data].xs(ticker, level="Ticker"))
ax_price.set_title(name)
ax_price.set_ylabel(data)
ax_price.grid(True)

ax_vol.bar(equities['MFI_14'].xs(ticker, level="Ticker").index, equities['MFI_14'].xs(ticker, level="Ticker"))
ax_vol.set_ylabel('MFI_14')
ax_vol.grid(True)

plt.tight_layout()
plt.show()
```

### Mapped Equities

```python
nan_mask = equities_mapped.isna().any(axis=1)
print(f'Tickers with at least one NaN: {nan_mask.sum()} / {len(equities_mapped)}')

nan_by_column = equities_mapped.isna().sum()
nan_by_column = nan_by_column[nan_by_column > 0]
print('\nNaN counts by column:')
print(nan_by_column)

equities_mapped[nan_mask]
```
