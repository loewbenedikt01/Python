## Parity

```python
def main():
    x = int(input('what is x? '))
    if is_even(x):
        print('Even')
    else:
        print('Odd')

def is_even(n):
    return n % 2 == 0

main()
```

## House

```python
name = input('what is your name? ')

if name == 'Harry' or name == 'Hermione' or name == 'Ron':
    print('Gryffindor')
elif name == 'Draco':
    print('Slytherin')
else:
    print('Who?')
```

```python
name = input('What is your name? ')

match name:
    case 'Harry' | 'Hermione' | 'Ron':
        print('Gryffindor')
    case 'Draco':
        print('Slytherin')
    case _:
        print('Who?')
```

# Loops

### While

```python
n = 3

while n != 0:
    print('meow')
    n = n - 1
```

```python
n = 1

while n <= 3:
    print('meow')
    n = n + 1
```

In programming, it is in general more used to count from 0 rather than 1

```python
n = 0

while n < 3:
    print('meow')
    n = n + 1
```

```python
n = 0
while n < 3:
    print('meow')
    n += 1
```

### For

```python
for n in [0, 1, 2]:
    print('meow')
```

Instead of a list, use a range

```python
for n in range(3):
    print('meow')
```

Use _ if you don't care about the variable later or mind at all, helps being pythonic

```python
for _ in range(3):
    print('meow')
```

```python
print('meow\n' * 3, end='')
```

```python
while True:
    n = int(input('what is n? '))
    if n > 0:
        break

for _ in range(n):
    print('meow')
```

```python
def main():
    number = get_number()
    meow(number)

def get_number():
    while True:
        n = int(input('what is n? '))
        if n > 0:
            break        
    return n

def meow(n):
    for _ in range(n):
        print('meow')

main()
```

# Exceptions

```python
try:
    x = int(input('what is x? '))
except ValueError:
    print('x is not an integer')
print(f'x is {x}')
```

```python
while True:
    try:
        x = int(input('what is x? '))
    except ValueError:
        print('x is not an integer')
    else:
        break

print(f'x is {x}')
```

### Code example

```python
def main():
    x = get_int()
    print(f'x is {x}')

def get_int():
    while True:
        try:
            x = int(input('what is x? '))
        except ValueError:
            print('x is not an integer')
        else:
            break
    return x

main()
```

### Code shortening and refinement

```python
def main():
    x = get_int()
    print(f'x is {x}')

def get_int():
    while True:
        try:
            x =  int(input('what is x? '))
            return x
        except ValueError:
            # print('x is not an integer')
            pass    # to not always print 'x is not an integer' 

main()
```
